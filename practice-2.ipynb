# Блок 3. Bootstrap и статистические гипотезы

## Задание 3.1. Bootstrap и доверительный интервал

```python
x = np.array([28, 31, 29, 45, 32, 30])
x.mean()
```

```python
B = 5000
bootstrap_means = np.empty(B)

np.random.seed(42)
for i in range(B):
    bootstrap_sample = np.random.choice(x, size=len(x), replace=True)
    bootstrap_means[i] = bootstrap_sample.mean()
```

Доверительный интервал

```python
np.quantile(bootstrap_means, 0.025)
```

```python
np.quantile(bootstrap_means, 0.975)
```

Вопросы для анализа
1. Почему Bootstrap используется именно с выборкой с возвращением?
> Чтобы сымитировать выборку из генеральной совокупности. Без возвращения мы бы просто переставляли те же значения, и разброс был бы неверным
2. Как наличие наблюдения 45 влияет на Bootstrap-распределение среднего?
> Это выброс он сдвигает bootstrap-распределение вправо и расширяет интервал
3. Почему результаты Bootstrap могут заметно изменяться при использовании очень
маленькой исходной выборки?
> Чем меньше n, тем сильнее каждый элемент влияет на результат, тем шире интервал

## Задание 3.3. p-value и принятие решения

После проведения A/B-теста статистический критерий дал:

p-value = 0.001

Уровень значимости:

α = 0.05

**Что необходимо сделать**

1. Дайте определение p-value.
> это вероятность наблюдаемых или более крайних данных при условии что нулевая гипотеза верна

2. Объясните, почему утверждение:

«p-value = 0.001 означает, что вероятность истинности нулевой гипотезы равна 0.1%» является неверным.
> p-value рассматривается при верности Н0 гипотезы, и не является ее вероятностью. Следовательно утверждение неверно.

3. Определите, является ли результат статистически значимым при $\alpha = 0.05$.
> Результат статистически значим, так как p-value строго меньше альфа, те данные странные для этой гипотезы
