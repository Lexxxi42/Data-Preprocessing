Медиана

```python
first_median = np.median(delivery_time)
first_median
```

Добавим еще наблюдение

```python
delivery_time_corrupted = np.append(delivery_time, 1000)
```

Среднее

```python
second_mean = np.mean(delivery_time_corrupted)
second_mean
```

Медиана

```python
second_median = np.median(delivery_time_corrupted)
second_median
```

Сравним изменения

```python
second_mean - first_mean
```

```python
second_median - first_median
```
